```python
n = 10
b = 2
binary = ''
while n > 0:
    rem = n % b
    binary = str(rem) + binary #0101
    n = n // b

print(binary)
```

```python
n = 10
b = 2
binary = ''
while n > 0:
    rem = n % b
    binary = binary + str(rem) #0101
    n = n // b

print(binary[::-1])
print(''.join(list(reversed(binary))))
```

```python
s = "swiss"
count = {}
for c in s:
    count[c] = count.setdefault(c,0)+1

for k,i in count.items(): #dictionary order may change approach with string order
    if i == 1:
        print("the first Non repeating word", k)
        break
    
```

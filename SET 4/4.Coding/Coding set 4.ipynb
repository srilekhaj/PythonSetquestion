````python
#1.Find the GCD of Two Numbers
Question: Write a Python program to find the Greatest Common Divisor (GCD) of 36 and 60.
````

```
GCD
36 = 2 X 2 X 3 X 3
60 = 2 X 2 X 3 X 5

COMMON FACTORS
2 X 2 X 3 = 12
```

````python
# Function to find HCF manually
def hcf(a, b):
    while b != 0:
        a, b = b, a % b  # Euclidean algorithm
    return a

# Function to find LCM manually
def lcm(a, b):
    return abs(a * b) // hcf(a, b)

# Input numbers
a = 36
b = 60

# Calculate manually
h = hcf(a, b)
l = lcm(a, b)

# Display results
print(f"HCF of {a} and {b} is: {h}")
print(f"LCM of {a} and {b} is: {l}")

````

````python
#2. Find the Sum of Digits of a Number
Question: Write a Python program to find the sum of the digits of the number 987.


````

````python
# method 1
res  = 0
n = 987
temp = n
while temp > 0 :
        
    rem = temp % 10
    res = res + rem
    temp = temp // 10
print("The sum of the digits :", res)
````

````python
#method 2
res = 0
for i in "987":
    res = res + int(i)

print(res)
````

````python
#4.Count the Frequency of Each Character in a String
Question: Write a Python program to count the frequency of each character in "programming".

````

````python
s= "programming"
counts = dict()

for ch in s:
    # print(ch)
    counts[ch]=counts.get(ch,0)+1

print(counts)
````
